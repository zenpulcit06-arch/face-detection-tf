```python
import os
os.environ['TF_ENABLE_ONEDNN_OPTS'] = '0'
os.environ['TF_CPP_MIN_LOG_LEVEL'] = '3'

import cv2 as cv
import numpy
import matplotlib.pyplot as plt
import tensorflow as tf
import tensorflow_datasets as tfds

print(f"{tf.__version__}")
print(f"{cv.__version__}")
```

```python
ldata, linfo = tfds.load("lfw",    split='train', with_info=True)
cdata, cinfo = tfds.load("cifar10", split='train', with_info=True)
```

```python
print(linfo.features)
print(cinfo.features)
```

```python
def preprocess_faces(dataset,label):
    image = tf.image.resize(dataset['image'],[64,64])/255
    return (image,label)

ldata = ldata.map(lambda x: preprocess_faces(x,1))
cdata = cdata.map(lambda x: preprocess_faces(x,0)).take(13000)
```

```python
for img, label in ldata.take(1):
    print('image shape: ', img.shape)
    print(f'Image min: {img.numpy().min()}, Image max: {img.numpy().max()}')
    print(f'labels: {label}')
    print(img.dtype)

print('for cdata')
for img, label in cdata.take(1):
    print('image shape: ', img.shape)
    print(f'Image min: {img.numpy().min()}, Image max: {img.numpy().max()}')
    print(f'labels: {label}')
    print(img.dtype)
```

```python
full_dataset = ldata.concatenate(cdata)
shuffled_data = full_dataset.shuffle(buffer_size = 26000)
```

```python
train_data, test_data = tf.keras.utils.split_dataset(shuffled_data,left_size = 0.8)
```

```python
train_data = train_data.batch(batch_size = 64)
test_data = test_data.batch(batch_size = 64)
train_data = train_data.prefetch(tf.data.AUTOTUNE)
test_data  = test_data.prefetch(tf.data.AUTOTUNE)
```

```python
inputs = tf.keras.Input(shape=(64, 64, 3))
x =  tf.keras.layers.Conv2D(32, kernel_size=3, activation='relu', padding='same')(inputs)
x =  tf.keras.layers.MaxPooling2D(pool_size=2)(x)

x = tf.keras.layers.Conv2D(64, kernel_size=3, activation='relu', padding='same')(x)
x = tf.keras.layers.MaxPooling2D(pool_size=2)(x)

x =  tf.keras.layers.Conv2D(128, kernel_size=3, activation='relu', padding='same')(x)
x = tf.keras.layers.MaxPooling2D(pool_size=2)(x)

x =  tf.keras.layers.Flatten()(x)

out_class = tf.keras.layers.Dense(1, activation='sigmoid')(x)

model = tf.keras.Model(inputs=inputs, outputs=out_class)

model.summary()
```

```python
model.compile(
    optimizer='adam',
    loss=['binary_crossentropy'],
    metrics=['accuracy']
)


history = model.fit(
    train_data,
    epochs = 5,
    validation_data=test_data
)
```

```python
def load_img(image_path,target_size=(64,64)):
    img_bgr = cv.imread(image_path)

    if img_bgr is None:
        raise FileNotFoundError(f"Could not find or open the image at: {image_path}")

    img_rgb = cv.cvtColor(img_bgr, cv.COLOR_BGR2RGB)

    img_resized = cv.resize(img_rgb, target_size)

    img_tensor = tf.convert_to_tensor(img_resized, dtype=tf.float32)

    img_tensor = img_tensor / 255.0

    img_tensor = tf.expand_dims(img_tensor, axis=0)

    return img_tensor

img = load_img("/mnt/c/Users/zenpu/OneDrive/Desktop/Python/face-detection-tf/closeup-face-portrait-of-pretty-blonde-girl-with-fresh-skin-horizontal-image-free-photo.jpg")
```
